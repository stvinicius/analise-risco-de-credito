## Como o algoritmo funciona

Três perguntas, nesta ordem:

1. **O que ele devolve?** Uma **nota de risco** de 0 a 1 — *não* “aprova/nega”.
2. **Como chega nessa nota?** Ou uma conta de pesos (logística), ou voto de árvores.
3. **O que o corte 50% faz?** Transforma a nota em 0/1 **só para a tabela**. O banco ainda não escolheu o corte (caderno 05).

---

### Dummy — o aluno que chuta “todo mundo é bom”

Como ~93 em cada 100 fichas são bom pagador, o chute mais comum é: **sempre 0**.

- Ele **não aponta nenhum** caloteiro.
- A fila dele é uma só nota para todo mundo → ordenar = cara ou coroa (**0,50**).
- Ele “acerta” ~93% — a mesma nota de quem aprova todo mundo.

Se a logística empatar com isso, o erro está no código (resposta invertida), não nos dados.

---

### Logística — uma conta de pesos (sem fórmula assustadora)

Imagine a ficha do cliente como uma **prova com várias questões** (idade, atraso, cartão…).

1. **Igualamos as unidades.** Idade 52 e uma bandeirinha 0/1 não se comparam. O computador deixa cada coluna com “média zero no treino”, para um ponto a mais significar a mesma coisa.
2. **Cada coluna ganha um peso.** Peso positivo = “isso aumenta o risco”. Peso negativo = “isso protege”.
3. **Somamos (valor × peso) de todas as colunas.** Sai um número qualquer: pode ser −2 ou +4. É só uma pontuação.
4. **Esprememos essa pontuação para 0–1** (uma curva em forma de S). Pontuação muito positiva → nota perto de 100%. Muito negativa → perto de 0%.

É isso. A fração dos livros só escreve o passo 4 em matemática. Você **não** precisa decorá-la para defender o projeto.

**O que é aquele “2,12” na seção 4?**  
Se o peso do cartão for +0,75, o computador também mostra **2,12**. Leia: “quando o cartão sobe um degrau típico do treino, as *chances relativas* de calote **multiplicam por cerca de 2**”. Não é “a taxa sobe 2 pontos percentuais”.

**O botão `C`:** controla se o aluno pode usar pesos **grandes**. `C` pequeno = pesos tímidos (menos decorar). `C` grande = pesos largos.

**Por que igualar unidades?** Sem isso, idade (número grande) “ganha” da bandeirinha 0/1 só porque o número é maior — e o freio dos pesos pune a coluna errada.

**Peso extra no caloteiro:** errar um caloteiro (~7% das fichas) dói ~14 vezes mais na caneta. As **notas** dos caloteiros sobem → no corte 50% você **aponta mais** deles. A **fila** (quem é pior que quem) quase não muda.

---

### Floresta — várias árvores votam

Cada árvore vê uma amostra sorteada das fichas e, em cada pergunta, só algumas colunas. A nota final é a **média** das notas das árvores.

- Enxerga **combinações** que a logística não vê (“jovem **e** cartão estourado **e** já atrasou 90+”).
- Não precisa igualar unidades: pergunta “cartão > 80%?”, não mede distâncias.
- Sem limite de profundidade, a árvore **decora** 120 mil fichas. Por isso começamos com profundidade 12 e no mínimo 20 pessoas em cada folha.

“Onde a floresta cortou” = em qual coluna ela mais perguntou. Variáveis com muitos valores (renda, dívida) saem infladas. **Não** é “essa coluna causa o calote” e **não** substitui a tabela do caderno 02.

---

### Boosting — árvores em fila, cada uma conserta a anterior

Na floresta as árvores trabalham **ao mesmo tempo** e votam. No boosting elas entram **uma depois da outra**: a segunda tenta acertar o que a primeira errou.

- Em planilha como a nossa, costuma **ordenar melhor** que a floresta no mesmo tempo de aula.
- Botões que importam: profundidade rasa, tamanho do passo, quantas árvores na fila.
- Não é XGBoost. É o boosting que já vem no sklearn — sem instalar nada.

---

### Por que o simulado interno fatia o treino *com a mesma % de calote*

Cinco fatias do **treino**. Cada uma herda ~6,7% de calote — a mesma ideia do corte do [03](03_data_preparation.ipynb).

Sem isso, uma fatia pode nascer com 4% de calote: a nota daquela fatia vira ruído e o botão “vencedor” é sorte. A nota do simulado é **a da fila** (ordenar), nunca o “acertei 93%” (o dummy ganharia).

**Gini** (na tabela) = duas vezes a nota da fila, menos 1. Zero = moeda; um = fila perfeita. O [05](05_evaluation.ipynb) desenha a curva.

2. **NaNs = 0.** O 03 imputou. Se aparecer NaN, você leu o raw por engano.
3. **Régua e flags (números desta run).** Mediana de renda **5400**, moda de dependentes **0**, P99 da util **~1,10**, DebtRatio **655** (com renda) / **~8138** (sem renda), P99 de renda **~23.018**.

   PD no teste: renda era missing **5,85%** < renda informada **6,89%** — mesmo sentido do 02 (5,61% vs 6,95% no arquivo inteiro). Se inverter, as flags foram criadas *depois* do `fillna` e o 03 precisa de replay.

## 1.3 As notas que vamos imprimir — leia isto antes do primeiro print

Imagine 30.000 clientes na prova. O aluno devolve 30.000 **notas de risco**. Aí fazemos **duas** coisas diferentes.

### A) A fila (não usa o corte 50%)

Pense numa fila: os mais arriscados na frente.

- **`auc_teste` (ordenar):** sorteio um caloteiro e um bom pagador. Em quantos % dos sorteios o caloteiro estava **na frente**? `0,50` = moeda. `1,00` = nunca erra a ordem. O [01](01_business_understanding.ipynb) chamou **acima de 0,80** de “bom”.
- **`gini_teste`:** a mesma fila, em outra escala (zero = moeda, um = perfeito).
- **`ap_teste`:** olha mais o **topo** da fila (os que o aluno mais acusou). Mais honesta quando só 7% são calote. O dummy vale **exatamente** 6,7%.

### B) O “sim/não” com corte provisório de 50%

Aqui a gente finge: “nota ≥ 50% → aponto como caloteiro”.

- **`recall_0.5` (apontar caloteiro):** de 100 pessoas que **deram** calote, quantas o aluno apontou? O 01 quer **mais de 75%** — mas *depois* do corte certo. Aqui o 50% é só para comparar alunos.
- **`acc_0.5` (acertei o sim/não):** impressa **só para mostrar que mente**. O dummy já faz 93%.

**Por que a logística sem peso extra costuma apontar poucos caloteiros no 50%:** ela percebe que calote é raro, então dá nota perto de **7%** para quase todo mundo. Quase ninguém passa de 50% — e “quantos caloteiros apontei” desaba. A **fila** pode estar ótima ao mesmo tempo. **Isso não é falha do 03.** É o corte 50% sendo uma régua ruim quando a coisa que importa é rara.

O simulado interno tira a média da fila em 5 fatias do **treino**. Usamos isso para *escolher* aluno/botão. A prova é um tiro só, no fim.



````python
# Mesmo objeto de CV em TODA busca/score: 5 fatias, embaralhadas, semente 42, ESTRATIFICADAS.
CV = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)


def metricas_teste(nome, modelo, X_te, y_te):
    """Uma linha da tabela. `predict_proba[:, 1]` = PD. O 0,5 é régua, não decisão."""
    proba = modelo.predict_proba(X_te)[:, 1]
    pred = (proba >= 0.5).astype(int)
    return {
        "modelo": nome,
        "auc_teste": roc_auc_score(y_te, proba),
        "ap_teste": average_precision_score(y_te, proba),
        "recall_0.5": recall_score(y_te, pred, zero_division=0),
        "bal_acc_0.5": balanced_accuracy_score(y_te, pred),
        "acc_0.5": float((pred == y_te).mean()),
        "gini_teste": 2 * roc_auc_score(y_te, proba) - 1,
    }


def cv_auc_treino(modelo, X, y):
    """Média ± desvio do AUC em 5 folds do TREINO. O teste não entra."""
    scores = cross_val_score(modelo, X, y, cv=CV, scoring="roc_auc", n_jobs=-1)
    return float(scores.mean()), float(scores.std()), scores


linhas = []  # cada modelo faz linhas.append(...); a seção 3.1 vira o quadro
````

## 2.1 Dummy — o chão (rode e confira os três números)

Antes de qualquer “AUC 0,86”, precisamos de um aluno que **não estudou**: chuta a classe mais comum.

O que *tem* que sair (senão pare):

| Métrica | Valor esperado | Por quê |
|---|---|---|
| `auc_teste` | **0,5** | Ranking constante não distingue ninguém |
| `recall_0.5` | **0** | Ninguém é chamado de caloteiro |
| `acc_0.5` | **~0,933** | 93,3% da prova é classe 0 |
| `ap_teste` | **~0,0668** | Average precision do classificador sem ranking = \(\pi\) do teste |
| `gini_teste` | **0** | \(2\times 0{,}5 - 1\) |

Se `auc_teste` **não** for 0,5 ou `recall_0.5` **não** for 0, pare: `y` invertido ou você leu `predict_proba` da classe 0.


````python
# most_frequent: ignora X e devolve sempre a classe mais comum do y_train (0).
dummy = DummyClassifier(strategy="most_frequent", random_state=42)
dummy.fit(X_train, y_train)
m = metricas_teste("dummy_most_frequent", dummy, X_test, y_test)
# CV de um constante é 0,5 por definição — não gastamos 5 folds nisso.
m["auc_cv_mean"], m["auc_cv_std"] = 0.5, 0.0
linhas.append(m)
print(pd.Series(m))
print("acurácia dummy (tem que ~0.933):", m["acc_0.5"])
````

### Como ler o print do dummy

Confira os três cadeados:

1. `auc_teste = 0.5` e `gini = 0` — ranking de moeda.
2. `recall_0.5 = 0` — zero caloteiro pego.
3. `acc_0.5 ≈ 0.933` — a armadilha do 01 em número.

Bônus: `ap_teste ≈ 0.0668` **é** a PD do teste. Average precision de um modelo sem ordem = prevalência. Qualquer modelo de verdade tem que **bater** esse 0,0668 no `ap_teste` (a logística vai a ~0,38).

Se passou: o `y_test` está do lado certo. Pode seguir.


## 2.2 Logística (baseline real) — o primeiro modelo que *aprende*

`Pipeline`: primeiro o scaler aprende média/desvio **só do treino** (ou do fold, na CV); depois a logística vê números na mesma escala.

Escolhas desta célula (propositalmente simples):

- `solver="lbfgs"`, `max_iter=500`, `C=1.0`
- **Sem** `class_weight`: otimiza a verossimilhança da amostra (~93% zeros)

O que esperar, se o 03 estiver certo:

- **Fila já na casa de 0,80–0,86** (o 01 chama isso de “bom”). A planilha tinha sinal; uma conta de pesos basta para ordenar.
- **Poucos caloteiros apontados no corte 50%** (nesta execução: ~17%). As notas ficam perto de 7%; quase ninguém cruza 50%. Aluno **tímido**, não aluno ruim.
- **“Acertos” ~94%**, um fio acima do dummy — e ainda assim inútil para o 01 (deixa ~83% dos caloteiros passarem neste corte).

**Interpretação depois (seção 4):** \(\beta>0\) aumenta odds de default. Esperado do 02: atrasos e utilização **positivos**; `age` e renda **negativos**.

A CV roda *antes* do `fit` final. Isso não “gasta” o teste: cada fold treina em 80% do *treino* e valida nos 20% restantes do *treino*.


````python
# Pipeline gruda o scaler no clf: na CV, cada fold refaz o scaler só com seus 4/5 do treino.
logreg = Pipeline(
    [
        ("scaler", StandardScaler()),
        (
            "clf",
            LogisticRegression(
                C=1.0,  # regularização "média"; a seção 3 testa outros C
                max_iter=500,
                solver="lbfgs",
                random_state=42,
            ),
        ),
    ]
)

mu, sd, _ = cv_auc_treino(logreg, X_train, y_train)  # 5 folds, SEM X_test
logreg.fit(X_train, y_train)  # fit final no treino inteiro
m = metricas_teste("logreg", logreg, X_test, y_test)  # um tiro no teste
m["auc_cv_mean"], m["auc_cv_std"] = mu, sd
linhas.append(m)
print("AUC CV 5-fold:", mu, "±", sd)
print(pd.Series({k: v for k, v in m.items() if k != "modelo"}))
````

### Como ler o print da logística (sem peso)

| Número | Valor | Como pensar |
|---|---|---|
| AUC-CV | 0,849 ± 0,006 | Estável entre folds (desvio pequeno). O 03 entregou sinal |
| AUC teste | **0,855** | Já é “bom” no 01 (> 0,80). A reta *ordena* risco |
| Caloteiros apontados no 50% | **16,6%** | Só 17 em cada 100 caloteiros levaram nota ≥ 50%. Tímido, não burro |
| Acc@0,5 | 0,937 | +0,4 p.p. vs dummy — e ainda deixa ~83% dos caloteiros passarem |
| AP | 0,377 | Muito acima do 0,067 do dummy: o ranking dos positivos funciona |

**Erro clássico agora:** “AUC 0,86 então o Recall também está bom”. Não. São perguntas diferentes. A próxima célula (`class_weight`) prova isso.


## 2.3 Logística com `class_weight="balanced"` — o primeiro modelo alinhado ao 01

Mesmo pipeline, mesma reta. Só muda a **conta de erro**: caloteiro pesa mais.

O que *deve* acontecer (é o experimento pedagógico desta célula):

| Métrica | vs 2.2 | Por quê |
|---|---|---|
| Fila | **quase igual** (sobe uns milésimos) | A ordem “quem é pior que quem” muda pouco |
| Caloteiros apontados no 50% | **sobe muito** (17% → ~76%) | As notas dos caloteiros sobem e cruzam os 50% |
| “Acertos” | **cai** (~94% → ~80%) | Você acusa mais gente boa. O 01 *aceita* isso para pegar mais caloteiro |

Isso ainda **não** é o threshold de negócio. É só o modelo deixando de ser tímido no corte provisório. O 05 é quem escolhe o corte com custo FN ≈ 20× FP.


````python
# Única diferença vs 2.2: class_weight. O 01 pede Recall — sem isso o corte 0,5 é cego.
logreg_bal = Pipeline(
    [
        ("scaler", StandardScaler()),
        (
            "clf",
            LogisticRegression(
                C=1.0,
                max_iter=500,
                solver="lbfgs",
                class_weight="balanced",  # peso ∝ 1/π ≈ 14× na classe 1
                random_state=42,
            ),
        ),
    ]
)

mu, sd, _ = cv_auc_treino(logreg_bal, X_train, y_train)
logreg_bal.fit(X_train, y_train)
m = metricas_teste("logreg_balanced", logreg_bal, X_test, y_test)
m["auc_cv_mean"], m["auc_cv_std"] = mu, sd
linhas.append(m)
print("AUC CV 5-fold:", mu, "±", sd)
print("caloteiros apontados no corte 50% (tem que SUBIR vs a tímida):", m["recall_0.5"])
print(pd.Series({k: v for k, v in m.items() if k != "modelo"}))
````

### Como ler o print da logística *balanced* (o experimento do 01)

Compare **lado a lado** com a célula anterior:

| | Sem peso | Com peso | O que mudou |
|---|---|---|---|
| AUC teste | 0,855 | **0,860** | Quase nada — o ranking é o mesmo aluno |
| Caloteiros apontados no 50% | 0,166 | **0,757** | 4,5× mais caloteiros cruzam o 50% |
| Acc@0,5 | 0,937 | **0,802** | Você “piorou” de propósito: mais FP, menos FN |

Essa é a frase que o 01 queria: **aceitar FP para reduzir FN**. O `class_weight` faz isso *sem* você escolher threshold no teste.

Ainda assim o 0,5 pode não ser o corte de custo 20×. O 05 mexe nisso.


## 2.4 Random Forest — o primeiro modelo que vê *interações*

Sem scaler: a árvore pergunta “util > 0,8?” e não se importa se idade está em anos ou em desvios.

Hiperparâmetros **já limitados** (não é grid — é cinto de segurança contra memorizar 120k linhas):

| Parâmetro | Valor | Intuição |
|---|---|---|
| `n_estimators` | 200 | Mais árvores = voto mais estável |
| `max_depth` | 12 | Sem teto, a árvore decora o treino |
| `min_samples_leaf` | 20 | Folha pequena = regra para 3 pessoas |
| `max_features` | `sqrt` | Cada corte vê ~4 das 13 colunas — árvores diferentes |
| `class_weight` | `balanced_subsample` | Repondera **em cada** bootstrap (melhor que um peso global único) |

CV aqui é mais lenta (minutos). `n_jobs=-1` usa todos os núcleos.

O que esperar: fila um pouco **acima** da logística (~0,86–0,87). Muitos caloteiros apontados por causa do peso, mas não é automaticamente o campeão — o boosting costuma passar na fila.


````python
# Sem Pipeline/scaler: árvore corta em limiares. n_jobs=-1 = todos os núcleos.
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,  # cinto: sem teto decora 120k linhas
    min_samples_leaf=20,
    max_features="sqrt",
    class_weight="balanced_subsample",  # repondera em CADA bootstrap
    random_state=42,
    n_jobs=-1,
)

mu, sd, _ = cv_auc_treino(rf, X_train, y_train)
rf.fit(X_train, y_train)
m = metricas_teste("random_forest", rf, X_test, y_test)
m["auc_cv_mean"], m["auc_cv_std"] = mu, sd
linhas.append(m)
print("AUC CV 5-fold:", mu, "±", sd)
print(pd.Series({k: v for k, v in m.items() if k != "modelo"}))
````

### Como ler o print da Random Forest

AUC-CV 0,862 / teste 0,867 — primeiro modelo a **passar** a logística no ranking. Interações (util alta *e* atraso, jovem *e* muitas linhas) que a reta não vê.

Caloteiros apontados no 50% = 73,8%: alto, mas **um pouco abaixo** dos 75% que o 01 pediu *neste corte*. Não jogue a floresta fora. No 05, uma linha 40% pode passar de 75% sem trocar de aluno.

Acc 0,826 > HGB 0,799: a floresta está um pouco mais “tímida” que o boosting no 0,5. Outro motivo para não escolher por acurácia.


## 2.5 HistGradientBoosting — boosting do sklearn (sem instalar XGBoost)

Equivalente prático a um LightGBM raso: árvores **fracas** em sequência, cada uma corrigindo o resíduo da anterior.

| Parâmetro | Valor | Intuição |
|---|---|---|
| `max_depth` | 6 | Árvores rasas (o “boost” vem da fila, não de uma árvore profunda) |
| `learning_rate` | 0,1 | Passo: menor = mais árvores para o mesmo fit, menos overfit |
| `max_iter` | 200 | Quantas árvores na fila |
| `min_samples_leaf` | 20 | Mesma ideia da RF |
| `l2_regularization` | 0,1 | Freio nos ganhos de cada folha |
| `class_weight` | `balanced` | Se o sklearn antigo não aceitar, caímos no `except` |

Nesta execução o boosting foi o **campeão** (fila 0,862 no simulado / 0,869 na prova / 77,8% dos caloteiros apontados no 50%) — passa o combinado do 01 **neste corte**. Isso **não** fecha o projeto: o 05 ainda escolhe a linha com o custo 20×.


````python
# Boosting raso do sklearn. try/except: versões velhas não têm class_weight no HGB.
hgb_params = dict(
    max_depth=6,
    learning_rate=0.1,
    max_iter=200,
    min_samples_leaf=20,
    l2_regularization=0.1,
    random_state=42,
)
try:
    hgb = HistGradientBoostingClassifier(class_weight="balanced", **hgb_params)
except TypeError:
    print("sklearn sem class_weight no HGB — treinando sem peso")
    hgb = HistGradientBoostingClassifier(**hgb_params)

mu, sd, _ = cv_auc_treino(hgb, X_train, y_train)
hgb.fit(X_train, y_train)
m = metricas_teste("hist_gb", hgb, X_test, y_test)
m["auc_cv_mean"], m["auc_cv_std"] = mu, sd
linhas.append(m)
print("AUC CV 5-fold:", mu, "±", sd)
print(pd.Series({k: v for k, v in m.items() if k != "modelo"}))
````

### Como ler o print do HistGB

| Número | Valor | Significado |
|---|---|---|
| AUC-CV | 0,862 | Melhor (ou empatado) ranking **dentro do treino** |
| AUC teste | 0,869 | Generalizou. Acima do 0,80 do 01 |
| Caloteiros apontados no 50% | **77,8%** | Pega ~78 em cada 100 — passa o 75% do 01 |
| Acc@0,5 | 0,799 | “Pior” que a logística tímida. **Bom sinal**: está acusando gente (FP) para não deixar FN |
| AP | 0,395 | ~6× o dummy (0,067) |
| Gini | 0,739 | \(2\times 0{,}869 - 1\) |

É o favorito de *ranking*. Ainda não é a política de crédito: falta o corte com custo 20×.

# 3. Hiperparâmetros (só no treino)

Busca **pequena** na logística balanced (`C`) — barata e didática.  
RF/HGB já nasceram com restrição de profundidade; um `RandomizedSearchCV` de 20×5 folds em 120k × RF seria aula de espera, não de crédito.

Como ler o ritual, passo a passo:

1. `param_distributions={"clf__C": np.logspace(-2, 2, 9)}` → nove candidatos de 0,01 a 100.
2. `scoring="roc_auc"` + `cv=CV` (5 folds **do treino**, estratificados).
3. `refit=True`: no fim o search faz `fit` no **treino inteiro** com o melhor `C`.
4. Só então medimos no teste — **um** tiro, não para escolher o `C`.

Nesta run o vencedor foi **`C=0,01`**, com AUC-CV **idêntica** à do `C=1,0` (0,8538). Lição: 13 features e 120k linhas — a logística **não** estava overfitting. Regularizar mais não muda o ranking. Ainda assim o ritual vale: você *demonstrou* que o `C` foi escolhido sem cola.


````python
# clf__C: o prefixo "clf__" aponta para o passo do Pipeline. 9 valores, de 0.01 a 100.
pipe_c = Pipeline(
    [
        ("scaler", StandardScaler()),
        (
            "clf",
            LogisticRegression(
                max_iter=500,
                solver="lbfgs",
                class_weight="balanced",
                random_state=42,
            ),
        ),
    ]
)

search = RandomizedSearchCV(
    pipe_c,
    param_distributions={"clf__C": np.logspace(-2, 2, 9)},
    n_iter=9,  # 9 candidatos × 5 folds = 45 fits — ainda é aula, não madrugada
    scoring="roc_auc",  # NUNCA accuracy (o dummy ganharia)
    cv=CV,  # só X_train
    random_state=42,
    n_jobs=-1,
    refit=True,  # no fim: fit no treino inteiro com o melhor C
)
search.fit(X_train, y_train)
logreg_tuned = search.best_estimator_

print("melhor C:", search.best_params_)
print("melhor AUC CV:", search.best_score_)

m = metricas_teste("logreg_balanced_C", logreg_tuned, X_test, y_test)
m["auc_cv_mean"] = float(search.best_score_)
m["auc_cv_std"] = float(
    pd.DataFrame(search.cv_results_)
    .query("rank_test_score == 1")["std_test_score"]
    .iloc[0]
)
linhas.append(m)
print(pd.Series({k: v for k, v in m.items() if k != "modelo"}))
````

### Como ler o print do `C`

`melhor C: 0.01` e `melhor AUC CV: 0.8538` — **igual** ao `C=1` da célula 2.3 (0,8538). O teste também empatou (0,8598).

Tradução: com 13 colunas e 120 mil linhas a logística **não precisa** de regularização forte nem fraca. O `C` não é o botão mágico deste problema. O botão que mudou a vida foi o `class_weight` (Recall 0,17 → 0,76), não o `C`.

Ainda assim gravamos `logreg_balanced_C`: o 05 pode querer o objeto cujo hiperparâmetro foi *documentado* via CV.

# 4. Interpretação (sem SHAP) — o modelo tem que *concordar* com o 02

SHAP está comentado no `requirements.txt`. Aqui usamos o que o sklearn já devolve:

- **\(\beta\)** da logística balanced → direção e tamanho (em desvios-padrão).
- **Gini importance** da RF → onde a floresta cortou (viesada a contínuas).

O [05](05_evaluation.ipynb) pode aprofundar com curvas. O teste desta seção é outro: se util e atraso 90+ não forem os maiores sinais, **desconfie do processed**, não “do algoritmo”.


## 4.1 Coeficientes da logística *balanced* — o gabarito do 02 em números

Features com scaler: \(\beta_j\) = efeito de **+1 desvio-padrão** no treino.  
`odds = exp(β)`. Um odds 2,12 quer dizer: “+1 DP nessa feature **multiplica** as odds de default por 2,12”, não “soma 2 pontos percentuais de PD”.

Confronte com a 6b. Nesta run (já executada):

| Feature | \(\beta\) | Odds | Bate com o 02? |
|---|---|---|---|
| `RevolvingUtilizationOfUnsecuredLines` | +0,75 | 2,12 | **Sim** — quintil alto tinha PD 18,5%; é o maior \(|\beta|\) |
| `NumberOfTimes90DaysLate` | +0,49 | 1,62 | **Sim** — PD 41% |
| `NumberOfTime30-59DaysPastDueNotWorse` | +0,39 | 1,48 | **Sim** — PD 20% |
| `NumberOfTime60-89DaysPastDueNotWorse` | +0,31 | 1,37 | **Sim**, mas \(\beta\) < 30–59. Os três atrasos **correlacionam**; a logística reparte o crédito. O 02 olha um de cada vez |
| `age` | −0,28 | 0,76 | **Sim** — idoso mais seguro |
| `MonthlyIncome` | −0,16 | 0,86 | **Sim** — renda alta protege (quando informada) |
| `DebtRatio` | −0,12 | 0,89 | **Parece estranho**, e é a lição do 03: dois mundos. Quem tem renda missing tem DebtRatio enorme **e** PD *menor* (5,61%). A reta pega esse atalho |
| `income_was_missing` | +0,04 | 1,04 | Quase zero. Depois de controlar DebtRatio/renda imputada, a flag sozinha quase não adiciona. Não apague ainda |
| `util_gt_10` | −0,05 | 0,95 | Quase zero. O cap já achatou a util; a flag é residual |
| `dependents_was_missing` | ≈ 0 | ≈ 1 | Esperado: poucos missings, sinal fraco no 02 |

O intercepto (~ −0,72) é a log-odds de um cliente **médio no treino** (todas as features = 0 *depois* do scaler). Não compare com a PD bruta 6,7% sem inverter a sigmoide — e mesmo assim o `class_weight` desloca esse nível.


````python
# named_steps["clf"] = a logística DENTRO do Pipeline (o scaler fica de fora desta conta).
clf = logreg_bal.named_steps["clf"]
coef = pd.Series(clf.coef_.ravel(), index=X_train.columns, name="beta")
tab = pd.DataFrame(
    {
        "beta": coef,
        "odds_ratio": np.exp(coef),  # e^β: multiplicador das odds por +1 DP
        "abs_beta": coef.abs(),
    }
).sort_values("abs_beta", ascending=False)
print("intercepto (log-odds base, features=0 após scale):", float(clf.intercept_[0]))
print()
print(tab.round(4).to_string())
````

### Como ler o print dos \(\beta\) (desta run)

Leia de cima para baixo, perguntando “o 02 concordaria?”:

1. **Util no topo (odds 2,12)** — sim. Quintil alto tinha PD 18,5%.
2. **90d, 30–59, 60–89 positivos** — sim. A ordem 90 > 30–59 > 60–89 **não** copia a PD univariada (41 / 20 / 36). A logística **reparte** sinal correlacionado: quem atrasa 90+ também atrasa 60–89, então o 60–89 “entrega” parte do crédito ao 90d.
3. **Idade negativa, renda negativa** — sim.
4. **DebtRatio negativa** — o 03 avisou. Missing de renda ⇒ DebtRatio gigante ⇒ PD *menor*. A reta usa o ratio como proxy de “não informou renda”.
5. **Flags ~0** — não é bug. O valor contínuo (já imputado/capado) já carrega o sinal.

Se util ou 90d saíssem **negativos**, pare: coluna trocada ou target invertido.

### Como saber se aprendeu (respostas, não só perguntas)

1. **Por que o dummy tem acurácia alta e Recall 0?**  
   Porque 93,3% das linhas são “bom pagador”. Chutar sempre 0 acerta 93% e **não pega nenhum** caloteiro. O 01 já tinha avisado que acurácia é armadilha.

2. **Por que scaler só na logística e dentro do `Pipeline`?**  
   A logística + regularização \(\ell_2\) compara *tamanho* de \(\beta\). Sem escala, idade (dezenas) “ganha” de uma flag 0/1. Árvore não usa distância. Dentro do Pipeline o scaler dá `fit` só no treino/fold — a média do teste não vaza.

3. **Por que “errar caloteiro dói mais” muda o apontar e quase não muda a fila?**  
   A fila olha a *ordem* das notas. O peso empurra as notas dos caloteiros para cima, então mais gente cruza os 50% — mas a ordem quase fica igual.

4. **Por que `RandomizedSearchCV` não pode ver `X_test`?**  
   Se o `C` vencedor for o que performa no teste, o teste deixou de ser prova. Vira mais um fold de treino. O 05 herdaria um número otimista.

5. **Por que importância Gini da RF não substitui a 6b do notebook 02?**  
   Gini premia variável contínua com muitos cortes. A 6b mede **PD real** em grupos que o negócio entende (“atrasou 90+ → 41%”). São perguntas diferentes: “onde a árvore cortou” vs “o que acontece no mundo”.

6. **Por que o 05 ainda precisa existir se a tabela do 04 já tem AUC?**  
   AUC não escolhe *quem aprova*. O banco precisa de um corte com custo FN/FP, de KS, de curva, de texto para stakeholder. O 04 compara *máquinas*; o 05 toma *decisão*.
